```python
import mysql.connector
import pandas as pd

# Suppress specific warnings from Pandas
import warnings
warnings.filterwarnings("ignore", category=UserWarning)

# Database connection
conn = mysql.connector.connect(
    user='ds160',
    password='SP2023',
    host='10.101.1.5',
    database='world'
)

#dont run this cell!
```

```python
# Fetch 'city' table
import warnings
warnings.filterwarnings("ignore", category=UserWarning)


city_query = "SELECT * FROM City;"
city_df = pd.read_sql(city_query, con=conn1)
city_df.head()
```

```python
# Fetch 'country' table
country_query = "SELECT * FROM Country;"
country_df = pd.read_sql(country_query, con=conn1)
country_df.sample(5)
```

```python

# Basic query: Filter cities with population > 500,000 and sort by Population in descending order
filtered_cities = city_df[city_df['Population'] > 500000].sort_values(by='Population', ascending=False)

# Display the first 5 rows
filtered_cities.head()
```

```python
# Merge 'city' and 'country' tables on 'CountryCode'
merged_df = pd.merge(
    city_df, country_df, 
    left_on='CountryCode', 
    right_on='Code'
)
merged_df.head()
```

```python

# Example query on the merged table: Cities with Population > 1M in Asia
filtered_merged = merged_df[
    (merged_df['Population_y'] > 1000000) & 
    (merged_df['Continent'] == 'Asia')
]
print("\nFiltered Merged Table (Cities with Population > 1M in Asia):")
filtered_merged.head()


```
